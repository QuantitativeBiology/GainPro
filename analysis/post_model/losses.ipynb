```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

# Config

```python
FIGSIZE = (6,4)
TRAIN_COLOR = "#fc8b64"
VAL_COLOR = "#909cc5"
```

# Plot functions

```python
def plot_generator_losses(
    train_generator_loss: list,
    val_generator_loss: list=None,
    name_loss: str="Generator Loss",
) -> None:
    epochs = np.arange(1, len(train_generator_loss) + 1)

    plt.figure(figsize=FIGSIZE)
    plt.xlabel('Epoch')
    plt.ylabel(name_loss)
    plt.plot(epochs, train_generator_loss, label="Train", color=TRAIN_COLOR)
    if val_generator_loss is not None:
        plt.plot(epochs, val_generator_loss, label="Validation", color=VAL_COLOR)
    plt.legend()
    plt.tick_params(axis='both', which='both', direction='out')
    ax = plt.gca()
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_axisbelow(True)


def plot_discriminator_losses(
    train_discriminator_loss: list,
    val_discriminator_loss: list=None,
    name_loss: str="Discriminator Loss",
) -> None:
    epochs = np.arange(1, len(train_discriminator_loss) + 1)

    plt.figure(figsize=FIGSIZE)
    plt.xlabel('Epoch')
    plt.ylabel(name_loss)
    plt.plot(epochs, train_discriminator_loss, label="Train", color=TRAIN_COLOR)
    if val_discriminator_loss is not None:
        plt.plot(epochs, val_discriminator_loss, label="Validation", color=VAL_COLOR)
    plt.legend()
    plt.tick_params(axis='both', which='both', direction='out')
    ax = plt.gca()
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_axisbelow(True)
```

```python
protogain_dir = Path("../../experiments/benchmark/PXD030304_no_control_multi_peptide_50pct_tissue/official/protogain/holdout/benchmark_2026-05-07_17:12:12/protogain/miss_10/run_001_seed_42_2026-05-07_17:12:12")
```

```python
training_metrics = pd.read_csv(protogain_dir / "training/metrics/train.csv")
```

```python
plot_generator_losses(training_metrics["generator_entropy"], name_loss="Discriminator Entropy")
plot_discriminator_losses(training_metrics["discriminator_loss"], name_loss="Discriminator Loss")
plot_generator_losses(training_metrics["generator_rmse"], name_loss="Generator RMSE")
```
